#Prerequisites


##Load 'user_001.csv' into Dataframe


```python
df = spark.read.csv(
    path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/user_dataset/users_001.csv",
    header=True,
    inferSchema=True,
)
display(df)
 
```

```python
df.write.save(
    format="PARQUET",
    path="/Volumes/wns24082026/quickstart_schema/sandbox/output_parquet",
)
```

```python
df_parquet = spark.read.parquet("/Volumes/wns24082026/quickstart_schema/sandbox/output_parquet")
display(df_parquet)
```

```python
df_parquet.createOrReplaceTempView("user_vw") #To execute a sql query on top of a dataframe
```

#[Disadvantage] -Updates are not allowed in parquet

```python
%sql
UPDATE
  user_vw
SET
  country = "Bharat"
where
  country = "India"
```

```python
%sql

SELECT * FROM user_vw limit 10
```


#[Disadvantage] - overwrite no rollback or version history

```python

#only the records where country is India will be written

from pyspark.sql.functions import col
df.filter(col("country")=="India").write.save(
    path = "/Volumes/wns24082026/quickstart_schema/sandbox/output_parquet",
    format = "PARQUET",
    mode = "overwrite"
)

df_parquet = spark.read.parquet("/Volumes/wns24082026/quickstart_schema/sandbox/output_parquet")
display(df_parquet)
```

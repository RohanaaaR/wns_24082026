```python
df = spark.read.csv(
    path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/employee.csv",
    inferSchema=True,
    header=True,
    sep="|",
    quote="'"

)
display(df)
```

```python
df.select("id","name").show()
```

```python
from pyspark.sql.functions import col

df.select(col("id").alias("userID") , col("name")).show()
```

#FILTERING

```python
df.filter((col("desig")=="Developer") | (col("desig")=="R&D Engineer")).show()
df.show()
```

```python
df.filter(col("desig").isin("Developer","R&D Engineer")).show()
df.show()
```


#Derive Columns

```python
from pyspark.sql.functions import lit
df.withColumn("Category" , lit("Not Defined")).show()
```

```python
df.printSchema()
```

```python
from pyspark.sql.functions import when

result_df = df.withColumn(
    "category",
    when(col("exp") >= 10, "Senior")
    .when(col("exp") >= 5, "Mid Level")
    .when(col("exp") > 0, "Junior")
    .otherwise("Invalid Experience")
)
```

```python
from pyspark.sql.functions import min, max, avg

result_df.groupBy("category").agg(min("exp"), max("exp"), avg("exp")).show()
```


#Dataframe write API

Write modes : 
    1.errorIfExist
    2.Overwrite
    3.APPEND
    4.IGNORE

```python
from pyspark.sql.types import *


#We use this to avoid extra stage
USER_SCHEMA = StructType(

    [
        StructField("id" , IntegerType()),
        StructField("age" , IntegerType()),
        StructField("gender" , StringType()),
        StructField("designation" , StringType())
    ]
)

df = spark.read.csv(
    path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/sample_dataset/users.csv",
    header=True,
    schema = USER_SCHEMA
)

df.write.save(
    format= "PARQUET",
    path= "/Volumes/wns24082026/quickstart_schema/sandbox/datasets/output/output_parquet",
    mode="IGNORE"

)
```
